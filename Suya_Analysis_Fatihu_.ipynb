
Find a 95% confidence interval for the average Total_NGN using a sample of 200 orders. Hint: average ± 1.96 × (standard deviation / √200).

```python
sample_200 = df_copy.sample(n=200, random_state=42)

sample_mean = sample_200['Total_NGN'].mean()
sample_std = sample_200['Total_NGN'].std()
n = len(sample_200)

standard_error = sample_std / np.sqrt(n)
margin_of_error = 1.96 * standard_error

lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error

print(f"Sample Mean (n=200): ₦{sample_mean:,.2f}")
print(f"Sample Standard Deviation: ₦{sample_std:,.2f}")
print(f"Margin of Error: ±₦{margin_of_error:,.2f}")
print(f"95% Confidence Interval: [₦{lower_bound:,.2f}, ₦{upper_bound:,.2f}]")
```
